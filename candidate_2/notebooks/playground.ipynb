## Aggregations

```python
products_df.columns = products_df.columns.str.lower()
stores_df.columns = stores_df.columns.str.lower()
```

```python
# sales_df, products_df, warranty_df, stores_df
sales_df.head(2)
```

```python
products_df.head(2)
```

```python
warranty_df.head(2)
```

```python
stores_df.head(2)
```

```python
df = sales_df.merge(products_df, on='product_id', how='left')
df.head(2)
```

```python
df = df.merge(warranty_df, on='sale_id', how='left')
df.head()
```

```python
df = df.merge(stores_df, on='store_id', how='left')
df.head(2)
```

```python
df.shape
```

```python
df.info()
```

```python
# Some transformations
df['revenue'] = df['quantity'] * df['price']
```

```python
# Revenue by Year
yearly_revenue = (
    df.groupby(df['sale_date'].dt.to_period('Y'))['revenue']
    .sum()
    .reset_index()
    .rename(columns={'sale_date': 'sale_month'})
)
yearly_revenue
```

```python
# Revenue by Product Category
category_revenue = (
    df.groupby('category_id')
    .agg(
        total_revenue= ('revenue', 'sum'), 
        total_units_sold = ('quantity', 'sum'), 
        transaction_count = ('sale_id', 'count')
    )
    .sort_values(by = 'total_revenue', ascending=False)
    .reset_index()
)
category_revenue
```

```python
# Top 3 Products by Volume
top_3_prods_by_vol = (
    df.groupby(['product_id', 'product_name'])
    .agg(total_units_sold= ('quantity', 'sum'))
    .sort_values(by = 'total_units_sold', ascending=False)
    .head(3)
    .reset_index()
)

top_3_prods_by_vol
```

```python
top_3_prods_revenue = (
    df.groupby(['product_id', 'product_name'])
    .agg(total_revenue = ('revenue', 'sum'))
    .sort_values(by = 'total_revenue', ascending=False)
    .head(3)
    .reset_index()
)

top_3_prods_revenue
```
